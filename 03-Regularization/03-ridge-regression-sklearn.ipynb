### Regularization using Sklearn

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import Ridge,Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x,y = load_diabetes(return_X_y=True, scaled=True)
```

### Using Ridge Regression

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=11)
```

```python
model = Ridge(alpha=0.1)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python
print("R2 score for Ridge Regression:",r2_score(y_test,y_pred))
```

### Using Lasso Regression

```python
model1 = Lasso(alpha=0.1)
model1.fit(x_train,y_train)
pred1 = model1.predict(x_test)
```

```python
print("R2 score for Ridge Regression:",r2_score(y_test,pred1))
```
