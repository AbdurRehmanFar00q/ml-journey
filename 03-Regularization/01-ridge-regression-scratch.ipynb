### Ridge Regression from strach

```python
import numpy as np

class RidgeRegressionStrach:
    def __init__(self,alpha=0.01):
        self.alpha = alpha
        self.coef_ = None
        self.intercept_ = None

    def fit(self,x,y):
        x = np.insert(x,0,1,axis=1)
        I = np.eye(x.shape[1])
        I[0,0] = 0  # Don't regularize the intercept
        betas = np.linalg.pinv(np.dot(x.T,x) + self.alpha * I ).dot(x.T).dot(y)
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]

    def predict(self,x):
        pred = np.dot(x,self.coef_) + self.intercept_   
        return pred
        
```

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x, y = load_diabetes(return_X_y=True, scaled=True)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=11)
```

```python
model = RidgeRegressionStrach(alpha=0.05)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python
print("R2 Score:", r2_score(y_test,y_pred))
```
