### Ridge Regression on Gradient descent

```python
import numpy as np

class RidgeRegressionGD:
    def __init__(self, lr=0.01, epoch=1000, alpha=0.01):
        self.lr = lr
        self.epoch = epoch
        self.alpha = alpha
        self.coeff_= None
        self.intercept_ = None
        

    def fit(self,x,y):
        self.coeff_ = np.ones(x.shape[1])
        self.intercept_ = 0

        for i in range(self.epoch):
            y_hat = np.dot(x, self.coeff_) + self.intercept_
            intercept_der = -2 * np.mean(y-y_hat)
            self.intercept_ = self.intercept_ - (self.lr * intercept_der)

            coeff_der = -2 * np.dot((y-y_hat),x)/x.shape[0] + (2 * self.alpha/x.shape[0]) * self.coeff_

            self.coeff_ = self.coeff_ - (self.lr * coeff_der)

    def predict(self,x):
        return self.intercept_ + np.dot(x, self.coeff_)
            
```

### Preapring the data

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x, y = load_diabetes(return_X_y=True, scaled=True)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=11)
```

### Making a model 

```python
model = RidgeRegressionGD(alpha=0.01, lr=0.1, epoch=10000)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python
print("R2 Score:", r2_score(y_test,y_pred))
```
