## Making own Linear Regression class


```python
import numpy as np

class LinearRegressionStrach:
    def __init__(self):
        self.coef_ = None
        self.intercept_ = None

    def fit(self,x,y):
        x = np.insert(x,0,1,axis=1)
        betas = np.linalg.pinv(np.dot(x.T,x)).dot(x.T).dot(y)
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]

    def predict(self,x):
        pred = np.dot(x,self.coef_) + self.intercept_   
        return pred
        


        
```

### Making a model
